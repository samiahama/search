```python
import json
import requests

# تحميل ground truth
with open('ground_truth.json', encoding='utf-8') as f:
    ground_truth = json.load(f)
```

```python
def precision_at_k(retrieved, relevant, k):
    retrieved_k = retrieved[:k]
    return len(set(retrieved_k) & set(relevant)) / max(len(retrieved_k), 1)

def recall_at_k(retrieved, relevant, k):
    retrieved_k = retrieved[:k]
    return len(set(retrieved_k) & set(relevant)) / max(len(relevant), 1)

def average_precision(retrieved, relevant):
    hits = 0
    sum_precisions = 0
    for i, doc_id in enumerate(retrieved, 1):
        if doc_id in relevant:
            hits += 1
            sum_precisions += hits / i
    return sum_precisions / max(len(relevant), 1)
```

```python
results = {}
for item in ground_truth:
    query = item["query"]
    params = {
        "query": query,
        "type": "tfidf",      # يمكنك تغييرها إلى "embedding" أو "hybrid"
        "dataset": "miracl_ar" # أو "msmarco"
    }
    response = requests.get("http://localhost:5000/api/search", params=params)
    docs = response.json()
    results[query] = [doc["doc_id"] for doc in docs]
```

```python
precisions = []
recalls = []
aps = []

for item in ground_truth:
    query = item["query"]
    relevant = item["relevant_docs"]
    retrieved = results.get(query, [])
    precisions.append(precision_at_k(retrieved, relevant, k=10))
    recalls.append(recall_at_k(retrieved, relevant, k=10))
    aps.append(average_precision(retrieved, relevant))

print("متوسط الدقة (Precision):", sum(precisions)/len(precisions))
print("متوسط الاسترجاع (Recall):", sum(recalls)/len(recalls))
print("متوسط متوسط الدقة (MAP):", sum(aps)/len(aps))
```
